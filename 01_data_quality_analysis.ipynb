```python
# Создаём копию исходного датасета
df_clean = df.copy()

print("Строк в исходных данных:", len(df))
print("Строк в копии:", len(df_clean))
```

```python
# Считаем полные дубликаты перед удалением
duplicates_before = df_clean.duplicated().sum()

# Удаляем повторяющиеся строки, сохраняя первое вхождение
df_clean = df_clean.drop_duplicates()

print("Удалено полных дубликатов:", duplicates_before)
print("Строк после удаления:", len(df_clean))
print("Полных дубликатов осталось:", df_clean.duplicated().sum())
```

```python
# Определяем строки, у которых номер счёта начинается с C
is_cancellation = (
    df_clean["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

# Проверяем отрицательное количество
is_negative_quantity = df_clean["Quantity"] < 0

print("Всего строк после удаления дубликатов:", len(df_clean))

print(
    "Строк со счетами на C:",
    is_cancellation.sum()
)

print(
    "Строк с отрицательным количеством:",
    is_negative_quantity.sum()
)

print(
    "Отрицательное количество без C:",
    (~is_cancellation & is_negative_quantity).sum()
)
```

```python
# Сумма по каждой строке
df_clean["LineAmount"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

# Определяем тип операции
invoice_text = df_clean["InvoiceNo"].astype(str)

df_clean["TransactionType"] = "Sale"

df_clean.loc[
    df_clean["Quantity"] < 0,
    "TransactionType"
] = "Return_or_cancellation"

df_clean.loc[
    invoice_text.str.startswith("C"),
    "TransactionType"
] = "Cancellation"

df_clean.loc[
    df_clean["UnitPrice"] == 0,
    "TransactionType"
] = "Zero_price"

df_clean.loc[
    df_clean["UnitPrice"] < 0,
    "TransactionType"
] = "Adjustment"

# Смотрим количество строк и сумму по каждой категории
transaction_summary = (
    df_clean.groupby("TransactionType")
    .agg(
        Rows=("InvoiceNo", "size"),
        TotalAmount=("LineAmount", "sum")
    )
    .sort_values("Rows", ascending=False)
)

display(transaction_summary.round(2))
```

```python
# Сумма по каждой строке
df_clean["LineAmount"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

# По умолчанию считаем строку обычной продажей
df_clean["TransactionType"] = "Sale"

invoice_text = df_clean["InvoiceNo"].astype(str)

# 1. Бухгалтерские корректировки с отрицательной ценой
df_clean.loc[
    df_clean["UnitPrice"] < 0,
    "TransactionType"
] = "Adjustment"

# 2. Отмены по номеру счёта, начинающемуся с C
df_clean.loc[
    invoice_text.str.startswith("C"),
    "TransactionType"
] = "Cancellation"

# 3. Отрицательное количество вне отмен
df_clean.loc[
    (df_clean["Quantity"] < 0)
    & (~invoice_text.str.startswith("C")),
    "TransactionType"
] = "Return"

# 4. Нулевая цена — для строк, ещё не отнесённых
# к отменам, возвратам или корректировкам
df_clean.loc[
    (df_clean["UnitPrice"] == 0)
    & (df_clean["Quantity"] >= 0)
    & (~invoice_text.str.startswith("C")),
    "TransactionType"
] = "Zero_price"

# Сводная таблица
transaction_summary = (
    df_clean.groupby("TransactionType")
    .agg(
        Rows=("InvoiceNo", "size"),
        TotalAmount=("LineAmount", "sum")
    )
    .sort_values("Rows", ascending=False)
)

display(transaction_summary.round(2))
```
